```python
falta = falta.copy()
falta["QtdManifestacoesRecebidas"] = pd.to_numeric(falta["QtdManifestacoesRecebidas"])
```

```python
falta["QtdManifestacoesRecebidas"].dtype
```

```python
mensal = falta.groupby("MesCompetencia")["QtdManifestacoesRecebidas"].sum()
mensal
```

```python
falta["MesCompetencia"] = pd.to_numeric(falta["MesCompetencia"])
```

```python
mensal = falta.groupby("MesCompetencia")["QtdManifestacoesRecebidas"].sum()
mensal
```

```python
mensal.plot(kind="bar")
plt.title("CPFL Paulista - Manifestações sobre falta de energia em 2025")
plt.xlabel("Mês")
plt.ylabel("Manifestações recebidas")
plt.show()

```

```python
aumento = (mensal.loc[9]-mensal.loc[8])/mensal.loc[8] * 100
print(f"De agosto para setembro, o aumento foi de {aumento:.2f} %")
```
